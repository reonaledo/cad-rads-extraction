```python
import pandas as pd
import numpy as np
from scipy.stats import chi2
from statsmodels.stats.contingency_tables import mcnemar
from statsmodels.stats.multitest import multipletests
from itertools import combinations
import seaborn as sns
import matplotlib.pyplot as plt

gpt4_result = make_predictions(gpt4_pred, labels)
gpt4o_result = make_predictions(gpt4o_pred, labels)
gpto1_result = make_predictions(gpto1_pred, labels)
claude_result = make_predictions(claude_pred, labels)
gemini_result = make_predictions(gemini_pred, labels)
deepseek_result = make_predictions(deepseek_pred, labels)

def prepare_results_data(category):
   
    # 모델별 성능
    models = {
        'GPT-4o': gpt4o_result[category].astype(int).values,
        'GPT-4': gpt4_result[category].astype(int).values,  
        'Claude-3.5-Sonnet': claude_result[category].astype(int).values,  
        'o1-mini': gpto1_result[category].astype(int).values,
        'Gemini-1.5-Pro': gemini_result[category].astype(int).values,
        'DeepSeek-14B': deepseek_result[category].astype(int).values
    }
    
    return pd.DataFrame(models)

def mcnemar_test_custom(model1_results, model2_results):
    # 2x2 table 생성
    both_correct = np.sum((model1_results == 1) & (model2_results == 1))
    b = np.sum((model1_results == 1) & (model2_results == 0))  # model1만 맞음
    c = np.sum((model1_results == 0) & (model2_results == 1))  # model2만 맞음
    both_wrong = np.sum((model1_results == 0) & (model2_results == 0))
    
    table = np.array([[both_correct, b], [c, both_wrong]])
    
    # b+c가 작으면 exact test 사용
    if b + c < 10:
        result = mcnemar(table, exact=True)
    else:
        result = mcnemar(table, exact=False, correction=True)
    
    print(f"Contingency table: b={b}, c={c}, b+c={b+c}")
    print(f"McNemar statistic: {result.statistic}, p-value: {result.pvalue}")
    
    return {
        'statistic': result.statistic,
        'pvalue': result.pvalue,
        'table': table,
        'b': b,
        'c': c
    }

def perform_pairwise_comparisons(results_df, component_name="stenosis_severity"):
    """
    모든 모델 pair에 대해 McNemar's test 수행
    """
    models = list(results_df.columns)
    comparison_results = []
    
    for model1, model2 in combinations(models, 2):
        test_result = mcnemar_test_custom(
            results_df[model1].values, 
            results_df[model2].values
        )
        
        comparison_results.append({
            'Model1': model1,
            'Model2': model2,
            'Component': component_name,
            'McNemar_statistic': test_result['statistic'],
            'p_value': test_result['pvalue'],
            'Model1_better': test_result['b'],  # Model1 correct, Model2 wrong
            'Model2_better': test_result['c'],  # Model1 wrong, Model2 correct
            'Significant': test_result['pvalue'] < 0.05
        })
    
    return pd.DataFrame(comparison_results)

def correct_multiple_comparisons(comparison_df, method='bonferroni'):
    """
    Multiple comparison correction 적용
    """
    p_values = comparison_df['p_value'].values
    
    if method == 'bonferroni':
        corrected_p = p_values * len(p_values)
        corrected_p = np.minimum(corrected_p, 1.0)
    elif method == 'fdr':
        # False Discovery Rate (Benjamini-Hochberg)
        reject, corrected_p, alpha_sidak, alpha_bonf = multipletests(p_values, method='fdr_bh')
    elif method == 'holm':
        reject, corrected_p, alpha_sidak, alpha_bonf = multipletests(p_values, method='holm')
    
    comparison_df['p_value_corrected'] = corrected_p
    comparison_df['Significant_corrected'] = corrected_p < 0.05
    
    return comparison_df

def create_significance_matrix(comparison_df, models):
    """
    모델 간 유의성 매트릭스 생성
    """
    n_models = len(models)
    matrix = np.zeros((n_models, n_models))
    
    model_to_idx = {model: i for i, model in enumerate(models)}
    
    for _, row in comparison_df.iterrows():
        i = model_to_idx[row['Model1']]
        j = model_to_idx[row['Model2']]
        
        if row['Significant_corrected']:
            if row['Model1_better'] > row['Model2_better']:
                matrix[i, j] = 1  # Model1 significantly better
                matrix[j, i] = -1  # Model2 significantly worse
            else:
                matrix[i, j] = -1  # Model1 significantly worse
                matrix[j, i] = 1  # Model2 significantly better
    
    return pd.DataFrame(matrix, index=models, columns=models)

def analyze_model_performance(category, true_df):
    """
    전체 분석 수행
    """
    print(f"Analyzing model performance for category: {category}")
    # 1. 데이터 준비 (실제 데이터로 교체 필요)
    results_df = prepare_results_data(category)
    
    # 마스킹 처리 추가
    if category == 'Plaque Burden':
        mask = (true_df['S'] == 0) & (true_df['CAC_available'] == 1)
        results_df = results_df[mask]
        print(f"Plaque Burden filtering applied: {len(results_df)} cases (S=0 & CAC_available=1)")
    elif category == 'CAD-RADS':
        mask = (true_df['N'] == 0)
        results_df = results_df[mask]
        print(f"CAD-RADS filtering applied: {len(results_df)} cases (N=0)")
    
    models = list(results_df.columns)
    
    print("=== Model Performance Analysis ===")
    print(f"Total cases: {len(results_df)}")
    print(f"Models analyzed: {', '.join(models)}")
    print()
    
    # 2. 기본 성능 통계 (마스킹 적용된 데이터로)
    print("Basic Performance Statistics:")
    performance_stats = results_df.mean().sort_values(ascending=False)
    for model, accuracy in performance_stats.items():
        print(f"{model}: {accuracy:.3f}")
    print()
    
    # 3. Pairwise comparisons
    print("=== Pairwise McNemar's Test Results ===")
    comparison_results = perform_pairwise_comparisons(results_df)
    
    # 4. Multiple comparison correction
    comparison_results = correct_multiple_comparisons(comparison_results, method='bonferroni')
    
    # 5. 결과 출력
    print("Significant differences (after Bonferroni correction):")
    significant_results = comparison_results[comparison_results['Significant_corrected']]
    
    if len(significant_results) > 0:
        for _, row in significant_results.iterrows():
            better_model = row['Model1'] if row['Model1_better'] > row['Model2_better'] else row['Model2']
            worse_model = row['Model2'] if row['Model1_better'] > row['Model2_better'] else row['Model1']
            print(f"{better_model} > {worse_model} (p = {row['p_value_corrected']:.3f})")
    else:
        print("No statistically significant differences found after correction.")
    
    print()
    
    # 6. 전체 순위
    print("=== Overall Model Ranking ===")
    # 각 모델이 다른 모델들보다 유의하게 좋은 횟수 계산
    win_counts = {}
    for model in models:
        wins = 0
        for _, row in significant_results.iterrows():
            if (row['Model1'] == model and row['Model1_better'] > row['Model2_better']) or \
               (row['Model2'] == model and row['Model2_better'] > row['Model1_better']):
                wins += 1
        win_counts[model] = wins
    
    ranked_models = sorted(win_counts.items(), key=lambda x: (-x[1], -performance_stats[x[0]]))
    
    for i, (model, wins) in enumerate(ranked_models, 1):
        print(f"{i}. {model} (Accuracy: {performance_stats[model]:.3f}, Significant wins: {wins})")
    
    return comparison_results, performance_stats, results_df

def visualize_results(comparison_results, performance_stats, results_df):
    """
    결과 시각화
    """
    models = list(results_df.columns)
    
    # 1. Performance comparison plot
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    performance_stats.plot(kind='bar')
    plt.title('Model Accuracy Comparison')
    plt.ylabel('Accuracy')
    plt.xticks(rotation=45)
    plt.tight_layout()
    
    # 2. Significance matrix heatmap
    plt.subplot(1, 3, 2)
    sig_matrix = create_significance_matrix(comparison_results, models)
    sns.heatmap(sig_matrix, annot=True, cmap='RdBu_r', center=0, 
                xticklabels=models, yticklabels=models)
    plt.title('Statistical Significance Matrix\n(1: Row > Col, -1: Row < Col)')
    
    # 3. P-value heatmap
    plt.subplot(1, 3, 3)
    p_matrix = np.ones((len(models), len(models)))
    model_to_idx = {model: i for i, model in enumerate(models)}
    
    for _, row in comparison_results.iterrows():
        i = model_to_idx[row['Model1']]
        j = model_to_idx[row['Model2']]
        p_matrix[i, j] = row['p_value_corrected']
        p_matrix[j, i] = row['p_value_corrected']
    
    sns.heatmap(pd.DataFrame(p_matrix, index=models, columns=models), 
                annot=True, cmap='viridis_r', 
                xticklabels=models, yticklabels=models)
    plt.title('Corrected P-values')
    
    plt.tight_layout()
    plt.show()
    
    return sig_matrix

# 실행
if __name__ == "__main__":
    comparison_results, performance_stats, results_df = analyze_model_performance('Plaque Burden', true_df=labels)
    sig_matrix = visualize_results(comparison_results, performance_stats, results_df)
    
    # 논문용 테이블 생성
    print("\n=== Table for Paper ===")
    paper_table = comparison_results[['Model1', 'Model2', 'p_value_corrected', 'Significant_corrected']].copy()
    paper_table['p_value_corrected'] = paper_table['p_value_corrected'].round(3)
    print(paper_table.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2
from statsmodels.stats.contingency_tables import mcnemar
from statsmodels.stats.multitest import multipletests
from itertools import combinations
import seaborn as sns
import matplotlib.pyplot as plt

gpt4_result = make_predictions(gpt4_pred, labels)
gpt4o_result = make_predictions(gpt4o_pred, labels)
gpto1_result = make_predictions(gpto1_pred, labels)
claude_result = make_predictions(claude_pred, labels)
gemini_result = make_predictions(gemini_pred, labels)
deepseek_result = make_predictions(deepseek_pred, labels)

def prepare_results_data(category):
   
    # 모델별 성능
    models = {
        'GPT-4o': gpt4o_result[category].astype(int).values,
        'GPT-4': gpt4_result[category].astype(int).values,  
        'Claude-3.5-Sonnet': claude_result[category].astype(int).values,  
        'o1-mini': gpto1_result[category].astype(int).values,
        'Gemini-1.5-Pro': gemini_result[category].astype(int).values,
        'DeepSeek-14B': deepseek_result[category].astype(int).values
    }
    
    return pd.DataFrame(models)

def mcnemar_test_custom(model1_results, model2_results):
    # 2x2 table 생성
    both_correct = np.sum((model1_results == 1) & (model2_results == 1))
    b = np.sum((model1_results == 1) & (model2_results == 0))  # model1만 맞음
    c = np.sum((model1_results == 0) & (model2_results == 1))  # model2만 맞음
    both_wrong = np.sum((model1_results == 0) & (model2_results == 0))
    
    table = np.array([[both_correct, b], [c, both_wrong]])
    
    # b+c가 작으면 exact test 사용
    if b + c < 10:
        result = mcnemar(table, exact=True)
    else:
        result = mcnemar(table, exact=False, correction=True)
    
    print(f"Contingency table: b={b}, c={c}, b+c={b+c}")
    print(f"McNemar statistic: {result.statistic}, p-value: {result.pvalue}")
    
    return {
        'statistic': result.statistic,
        'pvalue': result.pvalue,
        'table': table,
        'b': b,
        'c': c
    }

def perform_pairwise_comparisons(results_df, component_name="stenosis_severity"):
    """
    모든 모델 pair에 대해 McNemar's test 수행
    """
    models = list(results_df.columns)
    comparison_results = []
    
    for model1, model2 in combinations(models, 2):
        test_result = mcnemar_test_custom(
            results_df[model1].values, 
            results_df[model2].values
        )
        
        comparison_results.append({
            'Model1': model1,
            'Model2': model2,
            'Component': component_name,
            'McNemar_statistic': test_result['statistic'],
            'p_value': test_result['pvalue'],
            'Model1_better': test_result['b'],  # Model1 correct, Model2 wrong
            'Model2_better': test_result['c'],  # Model1 wrong, Model2 correct
            'Significant': test_result['pvalue'] < 0.05
        })
    
    return pd.DataFrame(comparison_results)

def correct_multiple_comparisons(comparison_df, method='bonferroni'):
    """
    Multiple comparison correction 적용
    """
    p_values = comparison_df['p_value'].values
    
    if method == 'bonferroni':
        corrected_p = p_values * len(p_values)
        corrected_p = np.minimum(corrected_p, 1.0)
    elif method == 'fdr':
        # False Discovery Rate (Benjamini-Hochberg)
        reject, corrected_p, alpha_sidak, alpha_bonf = multipletests(p_values, method='fdr_bh')
    elif method == 'holm':
        reject, corrected_p, alpha_sidak, alpha_bonf = multipletests(p_values, method='holm')
    
    comparison_df['p_value_corrected'] = corrected_p
    comparison_df['Significant_corrected'] = corrected_p < 0.05
    
    return comparison_df

def create_significance_matrix(comparison_df, models):
    """
    모델 간 유의성 매트릭스 생성
    """
    n_models = len(models)
    matrix = np.zeros((n_models, n_models))
    
    model_to_idx = {model: i for i, model in enumerate(models)}
    
    for _, row in comparison_df.iterrows():
        i = model_to_idx[row['Model1']]
        j = model_to_idx[row['Model2']]
        
        if row['Significant_corrected']:
            if row['Model1_better'] > row['Model2_better']:
                matrix[i, j] = 1  # Model1 significantly better
                matrix[j, i] = -1  # Model2 significantly worse
            else:
                matrix[i, j] = -1  # Model1 significantly worse
                matrix[j, i] = 1  # Model2 significantly better
    
    return pd.DataFrame(matrix, index=models, columns=models)

def analyze_model_performance(category, true_df):
    """
    전체 분석 수행
    """
    print(f"Analyzing model performance for category: {category}")
    # 1. 데이터 준비 (실제 데이터로 교체 필요)
    results_df = prepare_results_data(category)
    
    # 마스킹 처리 추가
    if category == 'Plaque Burden':
        mask = (true_df['S'] == 0) & (true_df['CAC_available'] == 1)
        results_df = results_df[mask]
        print(f"Plaque Burden filtering applied: {len(results_df)} cases (S=0 & CAC_available=1)")
    elif category == 'CAD-RADS':
        mask = (true_df['N'] == 0)
        results_df = results_df[mask]
        print(f"CAD-RADS filtering applied: {len(results_df)} cases (N=0)")
    
    models = list(results_df.columns)
    
    print("=== Model Performance Analysis ===")
    print(f"Total cases: {len(results_df)}")
    print(f"Models analyzed: {', '.join(models)}")
    print()
    
    # 2. 기본 성능 통계 (마스킹 적용된 데이터로)
    print("Basic Performance Statistics:")
    performance_stats = results_df.mean().sort_values(ascending=False)
    for model, accuracy in performance_stats.items():
        print(f"{model}: {accuracy:.3f}")
    print()
    
    # 3. Pairwise comparisons
    print("=== Pairwise McNemar's Test Results ===")
    comparison_results = perform_pairwise_comparisons(results_df)
    
    # 4. Multiple comparison correction
    comparison_results = correct_multiple_comparisons(comparison_results, method='bonferroni')
    
    # 5. 결과 출력
    print("Significant differences (after Bonferroni correction):")
    significant_results = comparison_results[comparison_results['Significant_corrected']]
    
    if len(significant_results) > 0:
        for _, row in significant_results.iterrows():
            better_model = row['Model1'] if row['Model1_better'] > row['Model2_better'] else row['Model2']
            worse_model = row['Model2'] if row['Model1_better'] > row['Model2_better'] else row['Model1']
            print(f"{better_model} > {worse_model} (p = {row['p_value_corrected']:.3f})")
    else:
        print("No statistically significant differences found after correction.")
    
    print()
    
    # 6. 전체 순위
    print("=== Overall Model Ranking ===")
    # 각 모델이 다른 모델들보다 유의하게 좋은 횟수 계산
    win_counts = {}
    for model in models:
        wins = 0
        for _, row in significant_results.iterrows():
            if (row['Model1'] == model and row['Model1_better'] > row['Model2_better']) or \
               (row['Model2'] == model and row['Model2_better'] > row['Model1_better']):
                wins += 1
        win_counts[model] = wins
    
    ranked_models = sorted(win_counts.items(), key=lambda x: (-x[1], -performance_stats[x[0]]))
    
    for i, (model, wins) in enumerate(ranked_models, 1):
        print(f"{i}. {model} (Accuracy: {performance_stats[model]:.3f}, Significant wins: {wins})")
    
    return comparison_results, performance_stats, results_df

def visualize_results(comparison_results, performance_stats, results_df):
    """
    결과 시각화
    """
    models = list(results_df.columns)
    
    # 1. Performance comparison plot
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    performance_stats.plot(kind='bar')
    plt.title('Model Accuracy Comparison')
    plt.ylabel('Accuracy')
    plt.xticks(rotation=45)
    plt.tight_layout()
    
    # 2. Significance matrix heatmap
    plt.subplot(1, 3, 2)
    sig_matrix = create_significance_matrix(comparison_results, models)
    sns.heatmap(sig_matrix, annot=True, cmap='RdBu_r', center=0, 
                xticklabels=models, yticklabels=models)
    plt.title('Statistical Significance Matrix\n(1: Row > Col, -1: Row < Col)')
    
    # 3. P-value heatmap
    plt.subplot(1, 3, 3)
    p_matrix = np.ones((len(models), len(models)))
    model_to_idx = {model: i for i, model in enumerate(models)}
    
    for _, row in comparison_results.iterrows():
        i = model_to_idx[row['Model1']]
        j = model_to_idx[row['Model2']]
        p_matrix[i, j] = row['p_value_corrected']
        p_matrix[j, i] = row['p_value_corrected']
    
    sns.heatmap(pd.DataFrame(p_matrix, index=models, columns=models), 
                annot=True, cmap='viridis_r', 
                xticklabels=models, yticklabels=models)
    plt.title('Corrected P-values')
    
    plt.tight_layout()
    plt.show()
    
    return sig_matrix

# 실행
if __name__ == "__main__":
    # comparison_results, performance_stats, results_df = analyze_model_performance('CAD-RADS', true_df=labels)
    # comparison_results, performance_stats, results_df = analyze_model_performance('Plaque Burden', true_df=labels)
    comparison_results, performance_stats, results_df = analyze_model_performance('HRP', true_df=labels)
    sig_matrix = visualize_results(comparison_results, performance_stats, results_df)
    
    # 논문용 테이블 생성
    print("\n=== Table for Paper ===")
    paper_table = comparison_results[['Model1', 'Model2', 'p_value_corrected', 'Significant_corrected']].copy()
    paper_table['p_value_corrected'] = paper_table['p_value_corrected'].round(3)
    print(paper_table.to_string(index=False))
```
